# 07_cnn：MLX 进阶

## 1. NumPy → MLX API 映射表
| NumPy / 手写概念 | MLX 写法 |
|---|---|
| `np.array` | `mlx.core.array` |
| 手写梯度 / 参数更新 | `mlx.nn.Conv2d` |
| NumPy 输出数组 | MLX lazy array，必要时 `mx.eval` |

## 2. MLX 等价实现
以下 cell 与主 Notebook 使用同一类输入形状，展示最小可运行 MLX 写法。

```python
try:
    import mlx.core as mx
    import mlx.nn as nn
    # MLX 的 Conv2d 封装了卷积核参数；这里关闭 bias，便于和手写卷积对照。
    conv = nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3, padding=1, bias=False)
    # MLX 示例使用 NHWC 布局：批量、高、宽、通道。
    x = mx.zeros((1, 8, 8, 1))
    y = conv(x)
    print('MLX 卷积输出形状：', y.shape, '；MLX Conv2d 默认示例使用 NHWC 数据布局。')
except (ImportError, RuntimeError) as e:
    print('提示：当前环境无法运行 MLX（未安装或无可用 Metal 设备）。Apple Silicon 用户可在本机可访问 GPU 的环境中运行。')
```

## 3. 数值一致性验证
安装 MLX 后，可把主 Notebook 中同一组 NumPy 输入复制到这里，断言输出误差 `< 1e-5`。本进阶 notebook 可跳过，不影响主线学习。

```python
import numpy as np
# 这里保留一个最小数值一致性检查模板：迁移到真实 MLX 输出时替换右侧数组即可。
np_value = np.array([1.0, 2.0, 3.0])
mlx_like_value = np.array([1.0, 2.0, 3.0])
assert np.allclose(np_value, mlx_like_value, atol=1e-5)
print('验证通过：数值一致性断言示例已通过；实际迁移时应替换为本模块 NumPy 与 MLX 的真实输出对照。')
```
